```python
# 필요한 모듈 읽어오기

import cv2
from google.colab.patches import cv2_imshow
```

```python
# PNG 이미지를 읽어 온다.
img = cv2.imread("gdrive/My Drive/CV/Basic Vision Skills/images/bukchon.PNG", cv2.IMREAD_COLOR)
print("shape: {}".format(img.shape))               # 형상 - (높이, 가로, 채널)
print("width: {} pixels".format(img.shape[1]))     # 넓이
print("height: {} pixels".format(img.shape[0]))    # 높이
print("channels: {}".format(img.shape[2]))         # 채널
```

```python
#  원본 이미지를 출력
cv2_imshow(img)
```

```python
# 원본 이미지를 JPG 형식 복사하여 저장한다.
cv2.imwrite("gdrive/My Drive/CV/Basic Vision Skills/images/bukchon.JPG", img)
```

```python
# JPG 이미지를 읽어 온다.
img2 = cv2.imread("gdrive/My Drive/CV/Basic Vision Skills/images/bukchon.JPG", cv2.IMREAD_COLOR)
print("shape: {}".format(img2.shape))               # 형상 - (높이, 가로, 채널)
print("width: {} pixels".format(img2.shape[1]))     # 넓이
print("height: {} pixels".format(img2.shape[0]))    # 높이
print("channels: {}".format(img2.shape[2]))         # 채널
```

```python
# JPG 이미지를 화면에 출력한다.
cv2_imshow(img2)
```

```python
# 원본 이미지의 색 공간 변경 : r, g, b 변수를 조합하여 RGB 색 공간으로 변경

# 각 채널을 분리
b, g, r = cv2.split(img)

# b, r을 서로 바꿔서 Merge
img3_rgb = cv2.merge([r, g, b])

# 이미지 형상 출력
print(img3_rgb.shape)   # 형상 - (496, 882, 3)
```

```python
# RGB 색 공간 변경 이미지를 출력한다.
cv2_imshow(img3_rgb)
```

```python
# 원본 이미지의 색 공간 변경 : b, g, r 변수를 조합하여 BGR 색 공간으로 변경

# b, r을 서로 바꿔서 Merge
img3_bgr = cv2.merge([b, g, r])

# 이미지 형상 출력
print(img3_bgr.shape)    # 형상 - (496, 882, 3)
```

```python
# BGR 색 공간 변경 이미지를 출력한다.
cv2_imshow(img3_bgr)
```
