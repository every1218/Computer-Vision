# 텐스플로 개념

### 텐서플로가 제공하는 MNIST(손글씨 이미지, 0~9 10개의 클래스, 28x28, 60000개 흑백 이미지)

```python
# 다음은 텐서플로가 제공하는 MNIST, Fashion-MNIST, Boston Housing, Reuters 네 가지 데이터 셋을 읽고 텐서의 구조를 확인

import tensorflow as tf
import tensorflow.keras.datasets as ds

# MNIST 읽고 텐서 모양 출력
(x_train, y_train),(x_test, y_test) = ds.mnist.load_data()
yy_train = tf.one_hot(y_train,10,dtype=tf.int8) # 원핫 코드로 변환
print("MNIST: ",x_train.shape,y_train.shape,yy_train.shape)  # MNIST: (60000, 28, 28) (60000,) (60000, 10)
```

### Fashion-MNIST

```python
# Fashion-MNIST 읽고 텐서 모양 출력
(x_train,y_train),(x_test,y_test) = ds.fashion_mnist.load_data()
yy_train = tf.one_hot(y_train,10,dtype=tf.int8)  # 원핫 코드로 변환
print("Fashion-MNIST: ",x_train.shape,y_train.shape,yy_train.shape) # Fashion-MNIST:(60000, 28, 28) (60000,) (60000, 10)
```

### Boston Housing(보스턴 시의 주택 가격 데이터)

```python
# Boston Housing 읽고 텐서 모양 출력
(x_train,y_train),(x_test,y_test) = ds.boston_housing.load_data()
print("Boston Housing: ",x_train.shape,y_train.shape)   # Boston Housing: (404, 13) (404,)
```

### Reuters(로이터 뉴스 데이터) 

```python
# Reuters 읽고 텐서 모양 출력
(x_train,y_train),(x_test,y_test) = ds.reuters.load_data()
print("Reuters: ",x_train.shape,y_train.shape)  # Reuters: (8982,) (8982,)
```
