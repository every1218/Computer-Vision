### 이미지 도구 OpenCV로 이미지 그려보기 

```python
# 파이썬에서 opencv-python 패키지 설치하기

#!pip3 install opencv-python
```

```python
import cv2
import numpy as np

with open(os.getcwd() + '/images/mandrill.png', 'rb') as f:
    file_bytes = np.asarray(bytearray(f.read()), dtype=np.uint8)
    img_color = cv2.imdecode(file_bytes, cv2.IMREAD_COLOR)
    img_gray = cv2.imdecode(file_bytes, cv2.IMREAD_GRAYSCALE)

cv2.imshow('color', img_color)
cv2.imshow('grayscale', img_gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# cv2 모듈에서 제공하는 imread() 함수를 통해 이미지 파일을 읽어 이미지 데이터를 생성할 수 있다.

# 다음은 OpenCV를 이용하여 이미지를 읽고 화면에 보이는 예제

import cv2 

img_gray = cv2.imread(os.getcwd()+'/images/mandrill.png', cv2.IMREAD_GRAYSCALE)  # 회색조 이미지 
img_color = cv2.imread(os.getcwd()+'/images/mandrill.png', cv2.IMREAD_COLOR)     # 컬러 이미지 

cv2.imshow('grayscale', img_gray)  # 맨드릴 원숭이 회색조 이미지 
cv2.imshow('color', img_color)     # 맨드릴 원숭이 컬러 이미지

# 다음 두 행은 키보드 입력을 기다렸다가 모든 창을 끄고 종료하는 코드이다.
cv2.waitKey(0)                     # 사용자 입력을 기다림 
cv2.destroyAllWindows()            # 모든 창을 없애고 프로그램 종료 
```

 ### OpenCV로 이미지 다루기  

```python
# 이미지에 선을 그릴 때는 cv2 모듈의 line() 함수를 사용할 수 있다.
# 이미지에 선을 그릴 때 arrowedLine() 함수는 끝지점에 화살표를 달아 준다.

#  다음은 OpenCV의 line(), arrowLine() 함수를 이용해 이미지에 선을 그리는 예제

import cv2 

img = cv2.imread(os.getcwd()+'/images/mandrill.png', cv2.IMREAD_COLOR)
cv2.line(img, (0,0), (200,200), (0,0,255), 5)          # 직선의 시작점과 끝점, 색상, 두께를 지정 
cv2.arrowedLine(img, (0,200), (200,20), (0,0,255), 5)  # 화살표의 시작점, 끝점, 색상, 두께지정
cv2.imshow('lined', img)
cv2.waitKey(0)                                         # 사용자 입력을 기다림 
cv2.destroyAllWindows()                                # 모든 창을 없애고 프로그램 종료 
```

```python
# rectangle() 함수를 호출하면 이미지에 사각형을 그려 준다. 
# putText() 함수를 이용하면 이미지에 글자를 출력할 수 있다.

# 다음은 OpenCV의 rectangle(), putText() 함수를 이용해 이미지에 사각형과 글자를 출력하는 예제

import cv2 

img = cv2.imread(os.getcwd()+'/images/mandrill.png', cv2.IMREAD_COLOR)
cv2.rectangle(img, (0,200), (200,20), (0,0,0), 5)                                # 사각형의 좌표, 색상을 지정 
cv2.putText(img, "rectline", (70,70), fontFace=2, fontScale=1, color=(0,250,0)) 
cv2.imshow('rectline', img)
cv2.waitKey(0)                                                                   # 사용자 입력을 기다림 
cv2.destroyAllWindows()                                                          # 모든 창을 없애고 프로그램 종료 
```

### 합성 사진 제작하기 

```python
# plt.imshow(X, cmap=None, norm=None, aspect=None, interpolation=None, alpha=None, vmin=None, vmax=None, origin=None,
#            extent=None, *, filternorm=True, filterrad=4.0, resample=None, url=None, data=None, **kwargs, )

# 두 개의 원본 이미지 출력해 보자.  

import matplotlib.pyplot as plt 

img_green_back = mpimg.imread(os.getcwd()+'/images/green_back.png')    
img_iceberg = mpimg.imread(os.getcwd()+'/images/iceberg.png')    

plt.figure(figsize=(10, 10))
plt.subplot(121)
plt.imshow(img_green_back)
plt.subplot(122)
plt.imshow(img_iceberg)
plt.show()
```

```python
# 이미지 두 개(image_a, image_b)를 섞어주는 함수는 addWeighted() 이다.
#  - image_merged = cv2.addWeighted(image_a, weight_for_a, image_b, weight_for_a)
#  - image_merged[x, y] 픽셀은 image_a[x, y] * weight_for_a + image_b[x, y] * weight_for_b의 값으로 설정  

# 다음은 OpenCV의 addWeighted() 함수를 이용해 두 개의 이미지를 합성하는 예제

import cv2 

img1 = cv2.imread(os.getcwd()+'/images/green_back.png', cv2.IMREAD_COLOR)
img2 = cv2.imread(os.getcwd()+'/images/iceberg.png', cv2.IMREAD_COLOR)
img1 = cv2.resize(img1, (300,400))    # 이미지를 300x400 크기로 설정
img2 = cv2.resize(img2, (300,400))    # 이미지를 300x400 크기로 설정 
x = 50                                # x값이 1로 설정 
weight = x / 100                      # x값이 0에서 100사이이므로 100으로 나누어 0에서 1사이 값으로 설정  

# image_merged[x, y] = image_a[x, y] * weight_for_a + image_b[x, y] * weight_for_b    
image_merged = cv2.addWeighted(img1, 1-weight, img2, weight, 0)
cv2.imshow('Display', image_merged)
cv2.waitKey(0)                        # 사용자 입력을 기다림 
cv2.destroyAllWindows()               # 모든 창을 없애고 프로그램 종료 
```

```python
# namedWindow() 함수를 이용하여 이름을 가진 창을 하나 만든다.
# createTrackbar() 함수로 창에 트랙바를 붙이면 된다.

# 다음은 OpenCV의 nameWindow(), createTrackbar() 함수를 이용해 두 개의 이미지를 합성하는 예제

import cv2 

global img1, img2                  # 두 이미지 변수를 전역 변수(프로그램 전체에서 사용할 수 있도록)로 설정  
img1 = cv2.imread(os.getcwd()+'/images/green_back.png', cv2.IMREAD_COLOR)
img2 = cv2.imread(os.getcwd()+'/images/iceberg.png', cv2.IMREAD_COLOR)
img1 = cv2.resize(img1, (300,400))  # 이미지를 300x400 크기로 설정
img2 = cv2.resize(img2, (300,400))  # 이미지를 300x400 크기로 설정 

def on_change_weight(x):
    weight = x / 100 
    img_merged = cv2.addWeighted(img1, 1-weight, img2, weight, 0)
    cv2.imshow('Display', img_merged)

cv2.namedWindow('Display') 
cv2.createTrackbar('weight', 'Display', 0, 100, on_change_weight)
cv2.waitKey(0)                      # 사용자 입력을 기다림 
cv2.destroyAllWindows()             # 모든 창을 없애고 프로그램 종료 
```
